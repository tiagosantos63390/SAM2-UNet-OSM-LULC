# Four-class auxiliary experiments

Compare four-class DT-MSE with mixed loss (building BCE+Dice). Channels are road=0, building=1, water=2, tree=3. Set all four mask directories below. See results/README.md for the historical mixed-loss runs.


```python
from pathlib import Path
import subprocess
import sys

REPO = Path.cwd()
if not (REPO / 'scripts/train.py').is_file():
    REPO = REPO.parent
assert (REPO / 'scripts/train.py').is_file(), 'Set REPO to the public repository checkout'
DATA_ROOT = Path('/path/to/OpenEarthMap_wo_xBD')
SAM_DIR = Path('/path/to/sam-checkpoints')
OSM_ROOT = Path('/path/to/OpenStreetMap')
WORK = Path('/path/to/new-run-directory')
SEEDS = [42, 43, 44]

COMMON = [sys.executable, str(REPO / 'scripts/train.py'),
          '--data-root', str(DATA_ROOT),
          '--train-txt', str(REPO / 'data/splits/train.txt'),
          '--val-txt', str(REPO / 'data/splits/val.txt'),
          '--sam-checkpoints-dir', str(SAM_DIR),
          '--osm-roads-dir', str(OSM_ROOT / 'roads/binary_masks'),
          '--osm-buildings-dir', str(OSM_ROOT / 'buildings/binary_masks'),
          '--seeds', *map(str, SEEDS)]
```

```python
subprocess.run(COMMON + ['--work-dir', str(WORK), '--validation-only',
                       '--osm-water-dir', str(OSM_ROOT / 'water/binary_masks'),
                       '--osm-trees-dir', str(OSM_ROOT / 'trees/binary_masks'),
                       '--experiments', 'four_class_L0p5_d6_dt_mse',
                       'four_class_L0p5_d6_mixed'], cwd=REPO, check=True)
```
